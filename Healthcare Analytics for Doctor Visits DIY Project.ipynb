# 10. Visualize the percentage of people getting govt health Insurance due to low income, due to old age and also the percentge of people having private health insurance

```python
# % of people getting govt health insurance due to low income

label = ['yes', 'no']

Y = df[df['freepoor'] == 'yes']
N = df[df['freepoor'] == 'no']

x = [Y.shape[0], N.shape[0]]

plt.figure(figsize=(5,5))
plt.pie(x, labels=label, autopct='%1.1f%%')
plt.title("% of people getting govt health insurance due to low income")
plt.show()


# % of people having private health insurance

Y = df[df['private'] == 'yes']
N = df[df['private'] == 'no']

x = [Y.shape[0], N.shape[0]]

plt.figure(figsize=(5,5))
plt.pie(x, labels=label, autopct='%1.1f%%')
plt.title("% of people having private health insurance")
plt.show()


# % of people getting govt health insurance due to old age,
# disability or veteran status

Y = df[df['freerepat'] == 'yes']
N = df[df['freerepat'] == 'no']

x = [Y.shape[0], N.shape[0]]

plt.figure(figsize=(5,5))
plt.pie(x, labels=label, autopct='%1.1f%%')
plt.title("% of people getting govt health insurance due to old age, disability or veteran status")
plt.show()
```

# 11. Plot a horizontal bar chart to analyze the reduced days of activity due to iilneess based on Gender

```python
# Group data by gender
db = df.groupby("gender")['reduced'].sum().to_frame().reset_index()

# Creating the bar chart
plt.figure(figsize=(8, 5))
plt.barh(db['gender'], db['reduced'])

# Adding the aesthetics
plt.title('Reduced Activity by Gender')
plt.xlabel("Gender")
# Show the plot
plt.show()
```
